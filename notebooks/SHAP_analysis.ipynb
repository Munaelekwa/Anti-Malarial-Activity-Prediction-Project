```python
# =========================
# Imports
# =========================
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
import joblib

# =========================
# Load trained RF model
# =========================
rf_model = joblib.load("../models/random_forest_model.pkl")

# =========================
# Load feature matrix
# =========================
X = np.load("../data/features_combined.npy")

# =========================
# Feature names
# =========================

descriptor_names = [
    'MolWt',
    'LogP',
    'TPSA',
    'NumHDonors',
    'NumHAcceptors',
    'NumRotatableBonds',
    'RingCount'
]

fp_names = [f'FP_{i}' for i in range(2048)]

feature_names = descriptor_names + fp_names

# Convert to dataframe
# Sample subset for SHAP
X_sample = pd.DataFrame(
    X[:1000],
    columns=feature_names
)

# =========================
# SHAP Explainer
# =========================
explainer = shap.TreeExplainer(rf_model)

# Calculate SHAP values
shap_values = explainer.shap_values(X_sample)

# For binary classification
# use class 1 (Active)
shap_class1 = shap_values[:, :, 1]

# =========================
# SHAP Summary Plot
# =========================
plt.figure()

shap.summary_plot(
    shap_class1,
    X_sample,
    max_display=20,
    show=False
)

plt.tight_layout()

# Save figure
plt.savefig(
    "../images/shap_summary_plot.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
